```python
import requests

def decode_secret_message(url):
    response = requests.get(url)
    html = response.text

    entries = []

    rows = html.split("<tr")

    for row in rows:
        cells = row.split("<td")

        if len(cells) < 4:
            continue

        char = cells[1].split(">")[1].split("<")[0].strip()
        x_str = cells[2].split(">")[1].split("<")[0].strip()
        y_str = cells[3].split(">")[1].split("<")[0].strip()

        if not x_str.isdigit():
            continue

        x = int(x_str)
        y = int(y_str)

        entries.append((char, x, y))

    max_x = max(e[1] for e in entries)
    max_y = max(e[2] for e in entries)

    grid = [[" "] * (max_x + 1) for _ in range(max_y + 1)]

    for char, x, y in entries:
        grid[y][x] = char

    for row in grid:
        print("".join(row))


url = "https://docs.google.com/document/d/e/2PACX-1vSvM5gDlNvt7npYHhp_XfsJvuntUhq184By5xO_pA4b_gCWeXb6dM6ZxwN8rE6S4ghUsCj2VKR21oEP/pub"
decode_secret_message(url)
```
