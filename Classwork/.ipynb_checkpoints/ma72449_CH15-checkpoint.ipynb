```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define parameters
h = 0.5 # step size
x_start, x_end = 0, 2
y0 = 1

num_steps = int((x_end - x_start) / h)
x_values = np.linspace(x_start, x_end, num_steps + 1)
y_values = np.zeros(num_steps + 1)

# Initial condition
y_values[0] = 1 # y(0) = 1


# Function representing dy/dx = -2y
def f(x, y):
    return -2 * y

# Implement Euler's method into a for loop
for h in np.arange(0.01, 1.01, 0.05):
    x = np.arange(x_start, x_end + h, h)
    y = np.zeros(len(x))
    y[0] = 1
    
    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])

# Plot the results
plt.plot(x_values, y_values, label="Euler's Method Approximation")
plt.xlabel('x')
plt.ylabel('y')
plt.title("Approximation of dy/dx = -2y using Euler's Method")
plt.legend()
plt.show()
```
