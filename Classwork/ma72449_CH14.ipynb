```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Define the function and the domain

f = lambda x: np.sin(x)
a, b = 0, 2 *np.pi #Interval
N = 50 #Number of sample points 

# Create a uniform grid 
x = np.linspace(a, b, N)
y = f(x)

# Plot the sampled points 
plt.plot(x, y, 'm-', label='Sampled Points')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Uniform Sampling of f(x) = sin(x)')
plt.legend()
plt.show()
```

```python
# Step size
h = (b - a) / (N - 1)

# Central Difference Approximation for the derivative
df_dx = (f(x + h) - f(x - h)) / (2 * h)

# Plot the approximation and the true derivative
plt.plot(x, np.cos(x), label='Exact Derivative')
plt.plot(x, df_dx, 'b--', label='Central Difference Approximation')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Approximating Derivative of f(x) = sin(x) using Central Difference')
```

```python
central = df_dx = (f(x + h) - f(x - h)) / (2 * h)
forward = (f(x + h) - f(x)) / h
backward = (f(x) - f(x-h)) / h
exact = np.cos(x)
```

```python
#plotting comparisons

plt.subplot(2, 2, 1)
plt.plot(x, exact, 'o-', color='pink')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title('Exact derivative of cos(x)')
plt.grid()

plt.subplot(2, 2, 2)
plt.plot(x, central, 'o-', color='pink')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title('Central Approximation')
plt.grid()

plt.subplot(2, 2, 3)
plt.plot(x, forward, 'o-', color='pink')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title('Forward Approximation')
plt.grid()

plt.subplot(2, 2, 4)
plt.plot(x, backward, 'o--', color='pink')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title('Backward Approximation')
plt.subplots_adjust(wspace=0.5, hspace=0.5)
plt.grid()

plt.show()
```
