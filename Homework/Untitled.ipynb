```python
import numpy as np
import matplotlib.pyplot as plt

# System of equations setup
# 3x1 - 0.1x2 - 0.2x3 = 7.85
# 0.1x1 + 7x2 - 0.3x3 = -19.3
# 0.3x1 - 0.2x2 + 10x3 = 71.4
A = np.array([[3.0, -0.1, -0.2],
              [0.1, 7.0, -0.3],
              [0.3, -0.2, 10.0]])
b = np.array([7.85, -19.3, 71.4])

# Calculate true solution for absolute error calculation
x_true = np.linalg.solve(A, b)

x0 = np.array([0.0, 0.0, 0.0]) # Initial guess
tolerances = [1e-3, 1e-6, 1e-9] # Specified tolerances

def jacobi(A, b, x0, tol, x_true):
    x = x0.copy()
    n = len(b)
    abs_errors = []
    iterations = 0
    while True:
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if i != j)
            x_new[i] = (b[i] - s) / A[i][i]
        
        # Absolute Error = |x_true - x_approx|
        error = np.linalg.norm(x_true - x_new, ord=np.inf)
        abs_errors.append(error)
        
        # Convergence check based on difference between iterations
        diff = np.linalg.norm(x_new - x, ord=np.inf)
        iterations += 1
        x = x_new
        if diff < tol or iterations > 100:
            break
    return x, iterations, abs_errors

def gauss_seidel(A, b, x0, tol, x_true):
    x = x0.copy()
    n = len(b)
    abs_errors = []
    iterations = 0
    while True:
        x_old = x.copy()
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if i != j)
            x[i] = (b[i] - s) / A[i][i]
        
        # Absolute Error = |x_true - x_approx|
        error = np.linalg.norm(x_true - x, ord=np.inf)
        abs_errors.append(error)
        
        # Convergence check based on difference between iterations
        diff = np.linalg.norm(x - x_old, ord=np.inf)
        iterations += 1
        if diff < tol or iterations > 100:
            break
    return x, iterations, abs_errors

# Plotting the Error Analysis for the strictest tolerance (10^-9)
plt.figure(figsize=(10, 6))

_, it_j, err_j = jacobi(A, b, x0, 1e-9, x_true)
_, it_gs, err_gs = gauss_seidel(A, b, x0, 1e-9, x_true)

plt.semilogy(range(1, it_j + 1), err_j, label='Jacobi Absolute Error', marker='o')
plt.semilogy(range(1, it_gs + 1), err_gs, label='Gauss-Seidel Absolute Error', marker='s')

plt.title('Error Analysis: $|x_{true} - x_{approx}|$ Convergence')
plt.xlabel('Iteration Number')
plt.ylabel('Absolute Error (Infinity Norm)')
plt.legend()
plt.grid(True, which="both", ls="-")
plt.show()

# Print iteration summary for different tolerances
for tol in tolerances:
    _, ij, _ = jacobi(A, b, x0, tol, x_true)
    _, igs, _ = gauss_seidel(A, b, x0, tol, x_true)
    print(f"Tol: {tol} | Jacobi iterations: {ij} | Gauss-Seidel iterations: {igs}")
```
