```python
#necessary imports
import numpy as np
import matplotlib.pyplot as plt
```

```python
#define known data points 
x_data = np.array([1, 2, 3, 4, 5])
y_data = np.array([2, 3, 5, 2, 4])

#incorporate Thomas algorithm
def thomas_algorithm(a, b, c, d):
    n = len(b)
    b_copy = b.copy().astype(float)
    d_copy = d.copy().astype(float)
    # Forward elimination
    for i in range(1, n):
        w = a[i-1] / b_copy[i-1]
        b_copy[i] = b_copy[i] - w * c[i-1]
        d_copy[i] = d_copy[i] - w * d_copy[i-1]
    # Back substitution
    x_sol = np.zeros(n)
    x_sol[-1] = d_copy[-1] / b_copy[-1]
    for i in range(n-2, -1, -1):
        x_sol[i] = (d_copy[i] - c[i] * x_sol[i+1]) / b_copy[i]
    return x_sol

# Step 3: Setup matrix function
def setup_tridiagonal_matrix(x, y):
    n = len(x)
    h = np.diff(x)  # Step 2: Calculate step sizes
    
    # Initialize the tridiagonal matrix coefficients for interior points
    B = np.zeros(n - 2)  # Main diagonal
    A = np.zeros(n - 3)  # Subdiagonal
    C = np.zeros(n - 3)  # Superdiagonal
    D = np.zeros(n - 2)  # Right-hand side vector
    
    # Populate coefficients based on spline continuity conditions
    for i in range(1, n - 1):
        idx = i - 1
        B[idx] = 2 * (h[i-1] + h[i])
        D[idx] = 6 * ((y[i+1] - y[i]) / h[i] - (y[i] - y[i-1]) / h[i-1])
        if idx > 0:
            A[idx-1] = h[i-1]
        if idx < n - 3:
            C[idx] = h[i]
            
    return A, B, C, D

# Step 4: Evaluate new spine points 
def evaluate_spline(x_eval, x, y, M, h):
    if x_eval <= x[0]: return y[0]
    if x_eval >= x[-1]: return y[-1]
    i = np.searchsorted(x, x_eval) - 1
    
    # Cubic spline formula using second derivatives (M)
    term1 = M[i] * (x[i+1] - x_eval)**3 / (6 * h[i])
    term2 = M[i+1] * (x_eval - x[i])**3 / (6 * h[i])
    term3 = (y[i]/h[i] - M[i]*h[i]/6) * (x[i+1] - x_eval)
    term4 = (y[i+1]/h[i] - M[i+1]*h[i]/6) * (x_eval - x[i])
    
    return term1 + term2 + term3 + term4

# 1. Setup the system
h_steps = np.diff(x_data)
a, b, c, d = setup_tridiagonal_matrix(x_data, y_data)

# 2. Solve for interior second derivatives (M)
M_interior = thomas_algorithm(a, b, c, d)

# 3. Apply Natural Spline Boundary Conditions (M1 = Mn = 0)
M = np.zeros(len(x_data))
M[1:-1] = M_interior

x_plot = np.arange(x_data[0], x_data[-1], 0.01)
y_plot = [evaluate_spline(val, x_data, y_data, M, h) for val in x_plot]

# Final Plot
plt.figure(figsize=(10, 6))
plt.plot(x_data, y_data, 'ro', markersize=8, label='Data Points')
plt.plot(x_plot, y_plot, 'b-', linewidth=2, label='Custom Cubic Spline')
plt.title('Homework: Cubic Spline Interpolation')
plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
