```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
def taylorAprx (n):

    x = np.arange(0,10.1, 0.1)
    for i in x:
        aprx = 0
    for z in range(n):
        numerator = ((-1)**z) * (i**(2*z + 1))
        denominator = math.factorial(2*z+1)
        aprx = aprx + (numerator/denominator) 
    


    exact = np.sin(i)
    error = np.abs(exact - aprx)
    return error
    

```

```python
aprx_one = taylorAprx(1)
aprx_two = taylorAprx(2)
aprx_three = taylorAprx(3)
aprx_four = taylorAprx(4)

print(aprx_one)
print(aprx_two)
print(aprx_three)
print(aprx_four)
```

```python
taylor_arr = [aprx_one, aprx_two, aprx_three, aprx_four]
plt.plot(taylor_arr)
```
