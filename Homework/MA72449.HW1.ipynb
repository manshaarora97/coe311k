```python
import math
import random 
import numpy as np 
import matplotlib.pyplot as plt 

def montecarlo_pi(number_of_shots):
    counter = 0
    number_of_hits = 0
    x = random.uniform(-1,1)
    y = random.uniform(-1,1)
    for i in range(0, number_of_shots):
        if(x**2 + y**2 < 1):
            number_of_hits = number_of_hits + 1;

        return 4*number_of_hits/number_of_shots
        
for i in range(0,100):
    calculated_pi = montecarlo_pi(1000)
    true_pi = np.pi
```

```python
import math
import numpy as np

error_plot = []

for i in range(0,100):
    calculated_pi = montecarlo_pi(1000)
    true_pi = np.pi
    true_error = abs(true_pi - calculated_pi)
    relative_error = abs((true_pi- calculated_pi)/(true_pi))
    x = relative_error
    error_plot.append(x)
    
    
print(true_error, relative_error)
plt.plot(error_plot)
plt.show()
```
